## 15.1 Correlation Matrix

A correlation matrix shows the pairwise relationships between numerical
variables in the dataset.

The correlation coefficient ranges from -1 to +1:

- +1 indicates a strong positive relationship.
- 0 indicates little or no linear relationship.
- -1 indicates a strong negative relationship.

The correlation matrix will help identify variables that may have
important relationships with agricultural performance.

```python
correlation_matrix = df[numerical_columns].corr()

correlation_matrix
```

## 15.2 Correlation Heatmap

A heatmap provides a visual representation of the correlation matrix.

It makes it easier to identify strong positive and negative relationships
among numerical variables.

```python
plt.figure(figsize=(16, 12))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0
)

plt.title("Correlation Heatmap of Numerical Variables")
plt.tight_layout()
plt.show()
```

## 15.3 Correlation with Crop Yield

Crop yield is one of the most important indicators of agricultural
performance.

Therefore, the correlation of yield with other numerical variables
will be examined to identify variables that show relatively stronger
positive or negative linear relationships with yield.

```python
yield_correlation = (
    correlation_matrix["Yield_Tonnes_Ha"]
    .sort_values(ascending=False)
)

yield_correlation
```

```python
print("Top positive correlations with Yield:")
print(
    yield_correlation
    .drop("Yield_Tonnes_Ha")
    .head(5)
)

print("\nTop negative correlations with Yield:")
print(
    yield_correlation
    .drop("Yield_Tonnes_Ha")
    .sort_values()
    .head(5)
)
```

## 15.4 Seasonal and Crop-Level Yield Analysis

Agricultural performance may differ not only across seasons but also
across different crops.

Therefore, average yield will be analyzed jointly by season and crop.

This multivariate view can help identify combinations of season and crop
that show relatively higher or lower average yield.

```python
season_crop_yield = pd.pivot_table(
    df,
    values="Yield_Tonnes_Ha",
    index="Season",
    columns="Crop",
    aggfunc="mean"
)

season_crop_yield
```

```python
plt.figure(figsize=(14, 7))

sns.heatmap(
    season_crop_yield,
    annot=True,
    fmt=".2f",
    cmap="YlGnBu"
)

plt.title("Average Yield by Season and Crop")
plt.xlabel("Crop")
plt.ylabel("Season")
plt.tight_layout()
plt.show()
```

## 15.5 Seasonal Irrigation and Yield Analysis

Therefore, average yield will be examined across both season and

## 16.6 Seasonal Profitability Risk

Average profit alone may not fully represent economic performance.

Therefore, the proportion of farms reporting negative profit will also
be examined across seasons.

A higher proportion of loss-making farms may indicate greater economic
risk or variability during a particular season.

```python
loss_analysis = df.groupby("Season").agg(
    Total_Farms=("Farm_ID", "count"),
    Loss_Making_Farms=("Profit_INR", lambda x: (x < 0).sum())
)

loss_analysis["Loss_Percentage"] = (
    loss_analysis["Loss_Making_Farms"]
    / loss_analysis["Total_Farms"]
    * 100
).round(2)

loss_analysis
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    data=loss_analysis.reset_index(),
    x="Season",
    y="Loss_Percentage"
)

plt.title("Percentage of Loss-Making Farms by Season")
plt.xlabel("Season")
plt.ylabel("Loss-Making Farms (%)")
plt.tight_layout()
plt.show()
```
